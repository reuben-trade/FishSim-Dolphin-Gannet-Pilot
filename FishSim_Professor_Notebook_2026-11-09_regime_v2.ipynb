## One fixed dolphin rule and fair controls

After H of continuous near-prey pressure, a dolphin chooses a physically feasible target, preferring fewer radius-2 neighbours, and makes one bounded strike before ordinary recovery. H does not certify completed herding; the fish need not be moving outward. No bird count, dive cue or capture bonus enters the dolphin controller. Gannets affect dolphins through prey dynamics and competition.

The native C++ FishSim engine is inherited from [yangyushi/FishSim](https://github.com/yangyushi/FishSim); this project adds predator controllers, experiment harnesses and analysis. Contact success q=.7 is uncalibrated conditional eligible-contact success, not success per observed field attack.

```python
# Enable after building the image as described in CURRENT_MODEL_SETUP.md.
RUN_CURRENT_MODEL = False
MODEL_BACKEND = "docker"
MODEL_GRID = "fine"
DEMO_DURATION_T = 0.5  # Choose 0.5 (startup) or 4 (short demonstration).

if RUN_CURRENT_MODEL:
    from datetime import datetime, timezone
    output = "outputs/current_demo_" + datetime.now(timezone.utc).strftime("%Y%m%d_%H%M%S")
    command = [sys.executable, "run_current_model.py", "--backend", MODEL_BACKEND,
               "--grid", MODEL_GRID, "--workers", "1", "--output", output]
    if DEMO_DURATION_T == 0.5:
        command += ["--quick"]
    elif DEMO_DURATION_T == 4:
        command += ["--run", "--duration", "4", "--school", "39064501", "--presences", "D", "DG"]
    else:
        raise ValueError("Choose a 0.5T or 4T demonstration.")
    subprocess.run(command, check=True, cwd=ROOT)
else:
    print("Model execution is disabled. Saved-result analysis remains available.")
```
